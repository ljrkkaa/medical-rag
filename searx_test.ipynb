```python
import pprint
from langchain_community.utilities import SearxSearchWrapper

SEARX_HOST = "http://127.0.0.1:8081"
DEFAULT_K = 5

search = SearxSearchWrapper(searx_host=SEARX_HOST, k=DEFAULT_K)

def show_results(query: str, **kwargs):
    results = search.results(query, **kwargs)
    pprint.pp(results)
    return results

search.run("深度学习", language="zh-CN", engines=["wiki"])
```

```python
results = show_results(
    "大型语言模型提示",
    num_results=5,
    categories="science",
    time_range="year",
)
```

```python
results = show_results(
    "大型语言模型提示",
    num_results=5,
    engines=["arxiv"],
)
```

```python
results = show_results(
    "large language model",
    num_results=20,
    engines=["github", "gitlab"],
)
```
